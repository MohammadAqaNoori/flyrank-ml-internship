## 2. The question: decision, action, cost of a wrong call

The decision is which pages should be reviewed first for a possible content refresh. A FlyRank editor or content strategist would use the ranking to prioritize limited review time. A wrong recommendation could waste editorial effort by reviewing a page that does not need attention, while a missed page could delay investigation of a meaningful decline. The model will therefore be treated as decision-support rather than an automatic refresh decision.

```python
# ML-02 — Section 2: Decision and action check

decision_supported = "Prioritize pages for content-refresh review"
human_actor = "Content editor or content strategist"
wrong_call_cost = "Wasted review effort or delayed investigation of a declining page"

print("Decision supported:", decision_supported)
print("Primary human actor:", human_actor)
print("Cost of a wrong call:", wrong_call_cost)
```
