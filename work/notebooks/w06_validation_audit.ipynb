## 1. Two findings + my methodology questions

### Finding 1 — Random Forest performance

The capstone reports that the Random Forest achieved 0.86 Precision@50 on the held-out test set.

The label comes from an observed impression decline between the previous-30-day and subsequent 30-day windows. The validation design supports this as a measured result on the evaluation sample.

Methodology question: because the split is a stratified random split rather than a time-aware backtest, does this result remain stable when the model is evaluated on a genuinely later time period?

### Finding 2 — Improvement over the baseline

The capstone reports that the Random Forest achieved 0.86 Precision@50 compared with 0.82 for the Week-4 baseline, an absolute improvement of 0.04.

Both methods use the same held-out test set and the same Precision@50 metric, which makes the comparison internally consistent.

Methodology question: would the observed improvement remain if the model were trained on earlier periods and evaluated strictly on a later period?

These questions do not invalidate the observed results. They identify what a stronger future validation design should test.

## 3. Leakage audit

The final feature set is checked against fields that could reveal the target or future outcome.

The target is constructed from the later 30-day impression window. Therefore future-window fields and the target itself must not appear among predictive features.

Pseudonymous identifiers are also excluded from predictive modeling because they identify records rather than provide meaningful pre-outcome signals.

```python
# ML-09 — Section 3: Leakage audit

final_features = [
    "impressions_prev30",
    "clicks_prev30",
    "ctr_prev30",
    "visible_query_count",
    "rare_query_count",
]

excluded_fields = {
    "trend_direction":
        "Label-derived outcome information",

    "trend_pct":
        "Label-derived outcome information",

    "impressions_last30":
        "Future outcome window",

    "clicks_last30":
        "Future outcome window",

    "client_hash_id":
        "Pseudonymous identifier, not a predictive feature",

    "content_hash_id":
        "Record identifier, not a predictive feature",

    "query_hash_id":
        "Record identifier, not a predictive feature",
}

leakage_audit = pd.DataFrame({
    "feature": final_features,

    "leakage_status": [
        "PASS",
        "PASS",
        "PASS",
        "PASS",
        "PASS",
    ],

    "reason": [
        "Historical previous-30-day signal",
        "Historical previous-30-day signal",
        "Calculated from previous-30-day signals",
        "Historical query visibility signal",
        "Historical query rarity signal",
    ]
})

display(leakage_audit)

print("\nExcluded fields:")
for field, reason in excluded_fields.items():
    print(f"- {field}: {reason}")
```

```python
# Explicitly document fields that are excluded.

excluded_fields = {
    "trend_direction": "Label-derived outcome information",
    "client_hash_id": "Pseudonymous identifier, not a predictive signal",
    "content_hash_id": "Record identifier, not a predictive signal",
    "query_hash_id": "Record identifier, not a predictive signal",
    "impressions_last30": "Future outcome window",
    "clicks_last30": "Future outcome window",
}

exclusion_audit = pd.DataFrame(
    [
        {
            "field": field,
            "reason": reason
        }
        for field, reason in excluded_fields.items()
    ]
)

display(exclusion_audit)
```

## 4. Claim rewrite

### Original strong claim

"The Random Forest predicts which pages will decline and can identify pages that should be refreshed."

### Safe research claim

"On this evaluation sample, the Random Forest ranked observed impression-decline cases with higher Precision@50 than the Week-4 baseline. The result is directional and supports a decision-support use case for prioritizing content review. It does not establish that the model will perform the same way on future time periods, that the model predicts Google's algorithm, or that refreshing a ranked page will recover traffic."

This wording separates the measured result from claims that the current validation design cannot establish.

```python
# ML-09 — Section 4: Claim-safety check

safe_claim_categories = [
    "observed",
    "measured",
    "directional",
    "decision-support",
]

unsupported_claims = [
    "causal proof",
    "predicting Google's algorithm",
    "guaranteed future performance",
    "proof that refreshing a page will recover traffic",
]

print("Supported claim language:")
for claim in safe_claim_categories:
    print("-", claim)

print("\nClaims explicitly avoided:")
for claim in unsupported_claims:
    print("-", claim)
```
